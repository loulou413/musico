# 00 — Dataset Exploration
Quick sanity check: load one track from each dataset and inspect annotations.

```python
import sys; sys.path.insert(0, '..')
import numpy as np
import librosa
import matplotlib.pyplot as plt
from src.preprocessing.common import load_config

cfg = load_config('../config/config.yaml')
print(cfg)
```

```python
# --- Saraga (Carnatic) ---
import mirdata
saraga = mirdata.initialize('saraga_carnatic', data_home=cfg['paths']['data_raw'] + '/saraga')
print('Saraga tracks:', len(saraga.track_ids))
t = saraga.track(saraga.track_ids[0])
print('  tala:', t.tala, '| raaga:', t.raaga)
```

```python
# Load audio + plot waveform
from src.preprocessing.saraga import load_saraga_track
track = load_saraga_track(t, data_home=cfg['paths']['data_raw'] + '/saraga')
plt.figure(figsize=(14, 2))
librosa.display.waveshow(track.audio, sr=track.sr)
plt.title(f"{track.track_id} ({track.metadata['tala']})")
plt.tight_layout()
```

```python
# Pitch contour
if track.f0_times is not None:
    voiced = track.f0_freqs > 0
    plt.figure(figsize=(14, 3))
    plt.plot(track.f0_times[voiced], track.f0_freqs[voiced])
    plt.xlabel('Time (s)'); plt.ylabel('Frequency (Hz)')
    plt.title('Reference pitch — Saraga')
    plt.tight_layout()
```

```python
# Beat positions
if track.beat_times is not None:
    print('First 10 beats (s):', track.beat_times[:10])
    print('First 5 sama (downbeat):', track.downbeat_times[:5] if track.downbeat_times is not None else 'N/A')
```
